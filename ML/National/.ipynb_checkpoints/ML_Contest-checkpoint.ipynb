````python
# ═══════════════════════════════════════════════════════════════
# SETUP — Run this cell first (do not modify)
# Contains utility functions and model classes from the workshop.
# ═══════════════════════════════════════════════════════════════
import numpy as np
import matplotlib.pyplot as plt
from collections import Counter

np.random.seed(42)

# ---- Utilities ----
def train_test_split(X, y, test_ratio=0.2, seed=None):
    rng = np.random.RandomState(seed)
    n = X.shape[0]
    idx = rng.permutation(n)
    n_test = int(n * test_ratio)
    test_idx = idx[:n_test]
    train_idx = idx[n_test:]
    return X[train_idx], X[test_idx], y[train_idx], y[test_idx]

def accuracy(y_true, y_pred):
    return np.mean(y_true == y_pred)

def mean_squared_error(y_true, y_pred):
    return np.mean((y_true - y_pred) ** 2)

def add_bias_column(X):
    n = X.shape[0]
    return np.hstack([np.ones((n, 1)), X])

def standardize(X_train, X_test=None):
    mean = X_train.mean(axis=0)
    std = X_train.std(axis=0)
    std[std == 0] = 1.0
    X_train_norm = (X_train - mean) / std
    if X_test is not None:
        X_test_norm = (X_test - mean) / std
        return X_train_norm, X_test_norm, mean, std
    return X_train_norm, mean, std

def sigmoid(z):
    z = np.clip(z, -500, 500)
    return np.where(z >= 0, 1/(1+np.exp(-z)), np.exp(z)/(1+np.exp(z)))

# ---- Softmax & One-hot ----
def softmax(Z):
    Z = Z - Z.max(axis=1, keepdims=True)
    exp_Z = np.exp(Z)
    return exp_Z / exp_Z.sum(axis=1, keepdims=True)

def one_hot(y, K):
    Y = np.zeros((len(y), K))
    Y[np.arange(len(y)), y] = 1
    return Y


# ---- Logistic Regression ----
class LogisticRegression:
    def __init__(self, learning_rate=0.1, n_epochs=500, l2_lambda=0.0):
        self.lr = learning_rate
        self.n_epochs = n_epochs
        self.l2_lambda = l2_lambda
        self.w = None
        self.cost_history = []
    def fit(self, X, y):
        Xb = add_bias_column(X)
        n, d = Xb.shape
        self.w = np.zeros(d)
        for epoch in range(self.n_epochs):
            h = sigmoid(Xb @ self.w)
            gradient = Xb.T @ (h - y) / n
            if self.l2_lambda > 0:
                reg = (self.l2_lambda / n) * self.w
                reg[0] = 0
                gradient += reg
            self.w -= self.lr * gradient
            eps = 1e-12
            cost = -np.mean(y * np.log(h + eps) + (1 - y) * np.log(1 - h + eps))
            self.cost_history.append(cost)
        return self
    def predict_proba(self, X):
        return sigmoid(add_bias_column(X) @ self.w)
    def predict(self, X):
        return (self.predict_proba(X) >= 0.5).astype(int)

# ---- Decision Tree Classifier ----
class DecisionTreeClassifier:
    def __init__(self, max_depth=5, min_samples_split=2, criterion='gini'):
        self.max_depth = max_depth
        self.min_samples_split = min_samples_split
        self.criterion = criterion
        self.tree = None
    def _gini(self, y):
        _, c = np.unique(y, return_counts=True)
        return 1 - np.sum((c / len(y)) ** 2)
    def _entropy(self, y):
        _, c = np.unique(y, return_counts=True)
        p = c / len(y)
        return -np.sum(p * np.log2(p + 1e-12))
    def _impurity(self, y):
        return self._gini(y) if self.criterion == 'gini' else self._entropy(y)
    def _best_split(self, X, y):
        n, d = X.shape
        best_gain = -float('inf')
        best_feat, best_thr = None, None
        parent_imp = self._impurity(y)
        for feat in range(d):
            for thr in np.unique(X[:, feat]):
                lm = X[:, feat] <= thr
                rm = ~lm
                nl, nr = lm.sum(), rm.sum()
                if nl == 0 or nr == 0: continue
                ci = (nl/n)*self._impurity(y[lm]) + (nr/n)*self._impurity(y[rm])
                gain = parent_imp - ci
                if gain > best_gain:
                    best_gain, best_feat, best_thr = gain, feat, thr
        return best_feat, best_thr, best_gain
    def _build(self, X, y, depth=0):
        n = len(y)
        if depth >= self.max_depth or n < self.min_samples_split or len(np.unique(y)) == 1:
            cls, cnt = np.unique(y, return_counts=True)
            return {'type': 'leaf', 'pred': cls[np.argmax(cnt)]}
        feat, thr, gain = self._best_split(X, y)
        if feat is None or gain <= 0:
            cls, cnt = np.unique(y, return_counts=True)
            return {'type': 'leaf', 'pred': cls[np.argmax(cnt)]}
        lm = X[:, feat] <= thr
        return {'type': 'node', 'feat': feat, 'thr': thr,
                'left': self._build(X[lm], y[lm], depth+1),
                'right': self._build(X[~lm], y[~lm], depth+1)}
    def fit(self, X, y):
        self.tree = self._build(X, y)
        return self
    def _predict_one(self, x, node):
        if node['type'] == 'leaf': return node['pred']
        if x[node['feat']] <= node['thr']:
            return self._predict_one(x, node['left'])
        return self._predict_one(x, node['right'])
    def predict(self, X):
        return np.array([self._predict_one(x, self.tree) for x in X])

# ---- Decision Tree Regressor ----
class DecisionTreeRegressor:
    def __init__(self, max_depth=5, min_samples_split=2):
        self.max_depth = max_depth
        self.min_samples_split = min_samples_split
        self.tree = None
    def _best_split(self, X, y):
        n, d = X.shape
        best_gain = -float('inf')
        best_feat, best_thr = None, None
        parent_var = np.var(y) if len(y) > 0 else 0
        for feat in range(d):
            for thr in np.unique(X[:, feat]):
                lm = X[:, feat] <= thr
                rm = ~lm
                nl, nr = lm.sum(), rm.sum()
                if nl == 0 or nr == 0: continue
                cv = (nl/n)*np.var(y[lm]) + (nr/n)*np.var(y[rm])
                gain = parent_var - cv
                if gain > best_gain:
                    best_gain, best_feat, best_thr = gain, feat, thr
        return best_feat, best_thr, best_gain
    def _build(self, X, y, depth=0):
        n = len(y)
        if depth >= self.max_depth or n < self.min_samples_split or len(y) <= 1:
            return {'type': 'leaf', 'pred': np.mean(y) if len(y) > 0 else 0}
        feat, thr, gain = self._best_split(X, y)
        if feat is None or gain <= 0:
            return {'type': 'leaf', 'pred': np.mean(y)}
        lm = X[:, feat] <= thr
        return {'type': 'node', 'feat': feat, 'thr': thr,
                'left': self._build(X[lm], y[lm], depth+1),
                'right': self._build(X[~lm], y[~lm], depth+1)}
    def fit(self, X, y):
        self.tree = self._build(X, y)
        return self
    def _predict_one(self, x, node):
        if node['type'] == 'leaf': return node['pred']
        if x[node['feat']] <= node['thr']:
            return self._predict_one(x, node['left'])
        return self._predict_one(x, node['right'])
    def predict(self, X):
        return np.array([self._predict_one(x, self.tree) for x in X])

# ---- Bagging ----
class BaggingClassifier:
    def __init__(self, n_estimators=20, max_depth=5, seed=42):
        self.n_estimators = n_estimators
        self.max_depth = max_depth
        self.rng = np.random.RandomState(seed)
        self.trees = []
    def fit(self, X, y):
        n = len(y)
        self.trees = []
        for _ in range(self.n_estimators):
            idx = self.rng.choice(n, size=n, replace=True)
            tree = DecisionTreeClassifier(max_depth=self.max_depth)
            tree.fit(X[idx], y[idx])
            self.trees.append(tree)
        return self
    def predict(self, X):
        preds = np.array([t.predict(X) for t in self.trees])
        return np.array([Counter(preds[:, i]).most_common(1)[0][0] for i in range(preds.shape[1])])

# ---- Random Forest ----
class RandomForestClassifierScratch:
    def __init__(self, n_estimators=30, max_depth=5, max_features=None, seed=42):
        self.n_estimators = n_estimators
        self.max_depth = max_depth
        self.max_features = max_features
        self.rng = np.random.RandomState(seed)
        self.trees = []
    def fit(self, X, y):
        n, d = X.shape
        if self.max_features is None:
            self.max_features = max(1, int(np.sqrt(d)))
        self.trees = []
        for _ in range(self.n_estimators):
            idx = self.rng.choice(n, size=n, replace=True)
            tree = self._build(X[idx], y[idx], 0)
            self.trees.append(tree)
        return self
    def _gini(self, y):
        if len(y) == 0: return 0
        _, c = np.unique(y, return_counts=True)
        return 1 - np.sum((c / len(y)) ** 2)
    def _build(self, X, y, depth):
        n, d = X.shape
        if depth >= self.max_depth or len(np.unique(y)) == 1 or n < 2:
            cls, cnt = np.unique(y, return_counts=True)
            return {'type': 'leaf', 'pred': cls[np.argmax(cnt)] if len(cls) > 0 else 0}
        feat_subset = self.rng.choice(d, size=min(self.max_features, d), replace=False)
        best_gain = -float('inf')
        best_feat, best_thr = None, None
        parent_gini = self._gini(y)
        for feat in feat_subset:
            for thr in np.unique(X[:, feat]):
                lm = X[:, feat] <= thr
                rm = ~lm
                nl, nr = lm.sum(), rm.sum()
                if nl == 0 or nr == 0: continue
                cg = (nl/n)*self._gini(y[lm]) + (nr/n)*self._gini(y[rm])
                gain = parent_gini - cg
                if gain > best_gain:
                    best_gain, best_feat, best_thr = gain, feat, thr
        if best_feat is None or best_gain <= 0:
            cls, cnt = np.unique(y, return_counts=True)
            return {'type': 'leaf', 'pred': cls[np.argmax(cnt)] if len(cls) > 0 else 0}
        lm = X[:, best_feat] <= best_thr
        return {'type': 'node', 'feat': best_feat, 'thr': best_thr,
                'left': self._build(X[lm], y[lm], depth+1),
                'right': self._build(X[~lm], y[~lm], depth+1)}
    def _predict_one(self, x, node):
        if node['type'] == 'leaf': return node['pred']
        if x[node['feat']] <= node['thr']:
            return self._predict_one(x, node['left'])
        return self._predict_one(x, node['right'])
    def predict(self, X):
        preds = np.array([[self._predict_one(x, t) for t in self.trees] for x in X])
        return np.array([Counter(preds[i]).most_common(1)[0][0] for i in range(len(X))])

# ---- AdaBoost ----
class DecisionStump:
    def __init__(self):
        self.feature = None; self.threshold = None; self.polarity = 1
    def fit(self, X, y, sample_weights=None):
        n, d = X.shape
        if sample_weights is None: sample_weights = np.ones(n) / n
        min_err = float('inf')
        for feat in range(d):
            for thr in np.unique(X[:, feat]):
                for pol in [1, -1]:
                    pred = np.ones(n)
                    if pol == 1: pred[X[:, feat] <= thr] = -1
                    else: pred[X[:, feat] > thr] = -1
                    err = np.sum(sample_weights[pred != y])
                    if err < min_err:
                        min_err = err
                        self.feature, self.threshold, self.polarity = feat, thr, pol
        return self
    def predict(self, X):
        pred = np.ones(X.shape[0])
        if self.polarity == 1:
            pred[X[:, self.feature] <= self.threshold] = -1
        else:
            pred[X[:, self.feature] > self.threshold] = -1
        return pred

class AdaBoost:
    def __init__(self, n_estimators=50, learning_rate=1.0):
        self.n_estimators = n_estimators
        self.lr = learning_rate
        self.stumps = []
        self.alphas = []
        self.training_errors = []
    def fit(self, X, y):
        n = len(y)
        w = np.ones(n) / n
        self.stumps = []; self.alphas = []
        for t in range(self.n_estimators):
            stump = DecisionStump()
            stump.fit(X, y, w)
            pred = stump.predict(X)
            err = np.sum(w[pred != y])
            err = max(min(err, 1 - 1e-10), 1e-10)
            alpha = 0.5 * np.log((1 - err) / err) * self.lr
            w = w * np.exp(-alpha * y * pred)
            w = w / np.sum(w)
            self.stumps.append(stump)
            self.alphas.append(alpha)
            train_pred = self.predict(X)
            self.training_errors.append(np.mean(train_pred != y))
        return self
    def predict(self, X):
        scores = np.zeros(X.shape[0])
        for alpha, stump in zip(self.alphas, self.stumps):
            scores += alpha * stump.predict(X)
        return np.sign(scores)

# ---- Gradient Boosting ----
class GradientBoostingClassifierScratch:
    def __init__(self, n_estimators=50, learning_rate=0.1, max_depth=3, seed=42):
        self.n_estimators = n_estimators
        self.lr = learning_rate
        self.max_depth = max_depth
        self.rng = np.random.RandomState(seed)
        self.trees = []
        self.init_logit = None
    def _sigmoid(self, z):
        return sigmoid(z)
    def fit(self, X, y):
        y01 = (y + 1) // 2
        self.init_logit = np.log(y01.mean() / (1 - y01.mean() + 1e-10) + 1e-10)
        F = np.full(len(y), self.init_logit)
        self.trees = []
        for t in range(self.n_estimators):
            p = self._sigmoid(F)
            residuals = y01 - p
            tree = DecisionTreeRegressor(max_depth=self.max_depth)
            tree.fit(X, residuals)
            update = tree.predict(X)
            F += self.lr * update
            self.trees.append(tree)
        return self
    def predict(self, X):
        F = np.full(X.shape[0], self.init_logit)
        for tree in self.trees:
            F += self.lr * tree.predict(X)
        return np.where(self._sigmoid(F) >= 0.5, 1, -1)

print("All dependencies loaded successfully.")
````

````python
# ═══════════════════════════════════════════════════════════════
# CONTEST DATASET — Run this cell (do not modify)
# ═══════════════════════════════════════════════════════════════
np.random.seed(42)

# ---- Contest Dataset (fixed seed, same for all students) ----
rng = np.random.RandomState(2024)

# Regression dataset
X_reg = rng.uniform(-4, 4, size=(120, 1))
y_reg_true = 0.5 * X_reg[:, 0]**2 - X_reg[:, 0] + 2
y_reg = y_reg_true + rng.randn(120) * 0.8

# Binary classification dataset (non-linearly separable)
n_cls = 200
t = np.linspace(0, 2 * np.pi, n_cls // 2)
X_cls = np.vstack([
    np.stack([np.cos(t) + rng.randn(n_cls//2)*0.15, np.sin(t) + rng.randn(n_cls//2)*0.15], axis=1),
    np.stack([0.5*np.cos(t) + rng.randn(n_cls//2)*0.15, 0.5*np.sin(t) + rng.randn(n_cls//2)*0.15], axis=1)
])
y_cls = np.array([0]*(n_cls//2) + [1]*(n_cls//2))
perm = rng.permutation(n_cls)
X_cls, y_cls = X_cls[perm], y_cls[perm]

# Train/test split for classification
X_cls_tr, X_cls_te, y_cls_tr, y_cls_te = train_test_split(X_cls, y_cls, test_ratio=0.25, seed=2024)

# Regression train/test split
X_reg_tr, X_reg_te, y_reg_tr, y_reg_te = train_test_split(X_reg, y_reg, test_ratio=0.25, seed=2024)

# Multi-class classification dataset (3 classes)
rng_mc = np.random.RandomState(2024)
npc = 80
X_mc = np.vstack([
    rng_mc.randn(npc, 2) + [3, 3],
    rng_mc.randn(npc, 2) + [-3, 3],
    rng_mc.randn(npc, 2) + [0, -3]
])
y_mc = np.array([0]*npc + [1]*npc + [2]*npc)
perm_mc = rng_mc.permutation(len(y_mc))
X_mc, y_mc = X_mc[perm_mc], y_mc[perm_mc]
X_mc_tr, X_mc_te, y_mc_tr, y_mc_te = train_test_split(X_mc, y_mc, test_ratio=0.25, seed=2024)

# High-dimensional regression dataset for regularization challenges
rng_hd = np.random.RandomState(2024)
n_hd = 100
d_hd = 20
X_hd = rng_hd.randn(n_hd, d_hd)
true_w_hd = np.zeros(d_hd)
true_w_hd[0] = 3.0; true_w_hd[3] = -2.0; true_w_hd[7] = 1.5  # only 3 relevant features
y_hd = X_hd @ true_w_hd + 0.5 + rng_hd.randn(n_hd) * 0.3
X_hd_tr, X_hd_te, y_hd_tr, y_hd_te = train_test_split(X_hd, y_hd, test_ratio=0.25, seed=2024)
# Standardize
X_hd_tr_s, X_hd_te_s, _, _ = standardize(X_hd_tr, X_hd_te)


print("Contest datasets loaded.")
print(f"Classification: {X_cls_tr.shape[0]} train, {X_cls_te.shape[0]} test")
print(f"Regression: {X_reg_tr.shape[0]} train, {X_reg_te.shape[0]} test")
print(f"Multi-class: {X_mc_tr.shape[0]} train, {X_mc_te.shape[0]} test (3 classes)")
print(f"High-dim regression: {X_hd_tr.shape[0]} train, {d_hd} features")

# Helper: autograder
def grade(answer, expected, tolerance=0.01, points=10, name=""):
    if isinstance(answer, (int, float, np.floating)):
        if abs(float(answer) - float(expected)) <= tolerance:
            print(f"  ✅ {name}: PASSED (+{points} pts)")
            return points
        else:
            print(f"  ❌ {name}: FAILED (got {answer:.4f}, expected ~{expected:.4f}, tolerance={tolerance})")
            return 0
    elif isinstance(answer, np.ndarray):
        if answer.shape == expected.shape and np.allclose(answer, expected, atol=tolerance):
            print(f"  ✅ {name}: PASSED (+{points} pts)")
            return points
        else:
            print(f"  ❌ {name}: FAILED")
            return 0
    else:
        if answer == expected:
            print(f"  ✅ {name}: PASSED (+{points} pts)")
            return points
        else:
            print(f"  ❌ {name}: FAILED (got {answer})")
            return 0

TOTAL_SCORE = 0
print("\nAutograder ready. Complete the challenges below and run the grading cells.")
````

### Challenge 2: Gradient Descent Convergence (15 pts)

**Task:** Implement gradient descent for linear regression **from scratch** (without using any pre-defined class). Use the contest regression dataset with polynomial features (degree 2, standardized).

**Requirements:**
- Initialize weights to zeros.
- Use learning rate $\eta = 0.01$ and run for exactly 1000 epochs.
- Use **full-batch** gradient descent (all training samples each step).
- Standardize the polynomial features (subtract mean, divide by std) before fitting.
- The gradient is: $\nabla = -\frac{2}{n}\mathbf{X}^T(\mathbf{y} - \mathbf{X}\mathbf{w})$
- Store the final training MSE in `c2_train_mse` and the final weight vector in `c2_w`.
- Store the loss after epoch 500 in `c2_loss_500`.

````python
# ═══ Challenge 2: Your code here ═══
# Build polynomial features
X_tr_p2 = np.hstack([X_reg_tr, X_reg_tr**2])
X_te_p2 = np.hstack([X_reg_te, X_reg_te**2])

# Standardize
mean_c2 = X_tr_p2.mean(axis=0)
std_c2 = X_tr_p2.std(axis=0)
std_c2[std_c2 == 0] = 1
X_tr_s = (X_tr_p2 - mean_c2) / std_c2
X_te_s = (X_te_p2 - mean_c2) / std_c2

# Add bias
X_tr_b = np.hstack([np.ones((len(X_tr_s), 1)), X_tr_s])

# Gradient descent
w_c2 = np.zeros(X_tr_b.shape[1])
lr_c2 = 0.01
losses_c2 = []
for epoch in range(1000):
    y_pred = X_tr_b @ w_c2
    grad = -2 / len(X_tr_b) * X_tr_b.T @ (y_reg_tr - y_pred)
    w_c2 = w_c2 - lr_c2 * grad
    loss = np.mean((y_reg_tr - y_pred)**2)
    losses_c2.append(loss)

c2_train_mse = losses_c2[-1]
c2_loss_500 = losses_c2[499]
c2_w = w_c2.copy()

print(f"Final train MSE: {c2_train_mse:.6f}")
print(f"Loss at epoch 500: {c2_loss_500:.6f}")
print(f"Final weights: {c2_w}")
````

````python
# ═══ Challenge 2: Grading ═══
print("Challenge 2: Gradient Descent Convergence")
TOTAL_SCORE += grade(c2_train_mse, 0.64, tolerance=0.15, points=5, name="Final train MSE")
TOTAL_SCORE += grade(c2_loss_500, 0.67, tolerance=0.2, points=5, name="Loss at epoch 500")
TOTAL_SCORE += grade(len(c2_w), 3, tolerance=0, points=5, name="Weight vector length (3)")
print()
````

### Challenge 3: Logistic Regression Decision Boundary (10 pts)

**Task:** Train a logistic regression model on the contest classification dataset (`X_cls_tr`, `y_cls_tr`). Then answer: **how many test samples are misclassified?**

**Requirements:**
- You may use the `LogisticRegression` class defined earlier in the notebook.
- Use `learning_rate=0.1` and `n_epochs=1000`.
- Do NOT standardize the features (use raw features).
- Store the number of misclassified test samples in `c3_n_misclassified`.

**Note:** The contest classification dataset is concentric circles — inherently non-linearly separable. This is intentional! Think about why a linear classifier struggles here and how many samples it gets wrong.

### Challenge 4: Decision Tree Split Selection (15 pts)

**Task:** Given the following tiny dataset, manually determine the **best Gini split** and answer questions about it.

```
Feature x1: [1, 2, 3, 4, 5, 6]
Feature x2: [5, 3, 4, 1, 2, 6]
Label:      [A, A, B, B, A, B]
```

**Requirements:**
- For each feature, consider every possible threshold (midpoints between consecutive sorted values).
- Use Gini impurity to determine the best split.
- Store the best feature index (0 for x1, 1 for x2) in `c4_best_feature`.
- Store the best threshold value in `c4_best_threshold` (use the midpoint).
- Store the Gini impurity *gain* (parent Gini - weighted child Gini) in `c4_best_gain`.

**Hint:** Parent Gini for 6 samples with 3×A and 3×B is $1 - (3/6)^2 - (3/6)^2 = 0.5$.

````python
# ═══ Challenge 4: Your code here ═══
X_c4 = np.array([[1, 5],
                  [2, 3],
                  [3, 4],
                  [4, 1],
                  [5, 2],
                  [6, 6]])
y_c4 = np.array([0, 0, 1, 1, 0, 1])  # 0=A, 1=B

def gini(y):
    _, c = np.unique(y, return_counts=True)
    return 1 - np.sum((c / len(y))**2)

parent_gini = gini(y_c4)
print(f"Parent Gini: {parent_gini:.4f}")

best_gain = -1
best_feature = -1
best_threshold = -1

for feat in range(2):
    sorted_vals = np.sort(np.unique(X_c4[:, feat]))
    for i in range(len(sorted_vals) - 1):
        threshold = (sorted_vals[i] + sorted_vals[i+1]) / 2
        left_mask = X_c4[:, feat] <= threshold
        right_mask = ~left_mask
        if left_mask.sum() == 0 or right_mask.sum() == 0:
            continue
        nl = left_mask.sum()
        nr = right_mask.sum()
        n = len(y_c4)
        child_gini = (nl/n) * gini(y_c4[left_mask]) + (nr/n) * gini(y_c4[right_mask])
        gain = parent_gini - child_gini
        print(f"  Feature {feat}, threshold={threshold:.1f}: left_gini={gini(y_c4[left_mask]):.4f}, "
              f"right_gini={gini(y_c4[right_mask]):.4f}, child_gini={child_gini:.4f}, gain={gain:.4f}")
        if gain > best_gain:
            best_gain = gain
            best_feature = feat
            best_threshold = threshold

c4_best_feature = int(best_feature)
c4_best_threshold = float(best_threshold)
c4_best_gain = float(best_gain)

print(f"\nBest split: feature={c4_best_feature}, threshold={c4_best_threshold:.2f}, gain={c4_best_gain:.4f}")
````

````python
# ═══ Challenge 4: Grading ═══
print("Challenge 4: Decision Tree Split Selection")
TOTAL_SCORE += grade(c4_best_feature, 0, tolerance=0, points=5, name="Best feature index")
TOTAL_SCORE += grade(c4_best_threshold, 2.5, tolerance=0.01, points=5, name="Best threshold")
TOTAL_SCORE += grade(c4_best_gain, 0.25, tolerance=0.01, points=5, name="Best Gini gain")
print()
````

### Challenge 6: Multi-class Logistic Regression with Regularization (20 pts)

**Task:** Implement **softmax (multi-class logistic) regression from scratch** with both L2 (Ridge) and L1 (Lasso) regularization. Train on the 3-class contest dataset and analyze how regularization affects the weights.

**Setup:** The dataset `X_mc_tr` (240 samples, 2 features, 3 classes) and `X_mc_te` are already prepared.

**Requirements:**

**Part A — Implementation (8 pts):**
- Implement a `SoftmaxRegression` class with:
  - `__init__(self, n_classes=3, learning_rate=0.1, n_epochs=1000, l2_lambda=0.0, l1_lambda=0.0)`
  - `fit(self, X, y)`: Add bias column, one-hot encode labels, gradient descent with regularization.
  - `predict(self, X)`: Return predicted class labels (integers).
- Use the `softmax` and `one_hot` functions provided in the Setup cell.
- **L2 gradient:** add `(l2_lambda / n) * W` to the weight gradient (do not regularize the bias row).
- **L1 gradient:** add `(l1_lambda / n) * sign(W)` to the weight gradient (do not regularize the bias row).
- Store the class in `SoftmaxRegression`.

**Part B — No regularization (3 pts):**
- Train with `l2_lambda=0, l1_lambda=0, learning_rate=0.5, n_epochs=1000`.
- Store test accuracy in `c6_test_acc_none`.
- Store the L2 norm of all weights (excluding bias) in `c6_weight_norm_none`.

**Part C — Ridge (L2) (3 pts):**
- Train with `l2_lambda=10.0, learning_rate=0.5, n_epochs=1000`.
- Store test accuracy in `c6_test_acc_l2`.
- Store the L2 norm of all weights (excluding bias) in `c6_weight_norm_l2`.

**Part D — Lasso (L1) (3 pts):**
- Train with `l1_lambda=10.0, learning_rate=0.5, n_epochs=1000`.
- Store test accuracy in `c6_test_acc_l1`.
- Store the number of weights (excluding bias) with absolute value < 1e-4 in `c6_n_zero_weights_l1`.

**Part E — Comparison (3 pts):**
- Store `True` or `False` in `c6_l2_reduced_norm` — did L2 regularization reduce the weight norm compared to no regularization?
- Store `True` or `False` in `c6_l1_created_zeros` — did L1 regularization create zero weights (count > 0)?

````python
# ═══ Challenge 6: Your code here ═══

class SoftmaxRegression:
    def __init__(self, n_classes=3, learning_rate=0.1, n_epochs=1000, l2_lambda=0.0, l1_lambda=0.0):
        self.n_classes = n_classes
        self.lr = learning_rate
        self.n_epochs = n_epochs
        self.l2_lambda = l2_lambda
        self.l1_lambda = l1_lambda
        self.W = None

    def fit(self, X, y):
        Xb = add_bias_column(X)
        n, d = Xb.shape
        K = self.n_classes
        Y = one_hot(y, K)
        self.W = np.zeros((d, K))
        for epoch in range(self.n_epochs):
            logits = Xb @ self.W
            probs = softmax(logits)
            grad = Xb.T @ (probs - Y) / n
            if self.l2_lambda > 0:
                reg_l2 = (self.l2_lambda / n) * self.W
                reg_l2[0, :] = 0  # don't regularize bias
                grad += reg_l2
            if self.l1_lambda > 0:
                reg_l1 = (self.l1_lambda / n) * np.sign(self.W)
                reg_l1[0, :] = 0  # don't regularize bias
                grad += reg_l1
            self.W -= self.lr * grad
        return self

    def predict(self, X):
        Xb = add_bias_column(X)
        return np.argmax(Xb @ self.W, axis=1)

# Part B: No regularization
sm_none = SoftmaxRegression(n_classes=3, learning_rate=0.5, n_epochs=1000, l2_lambda=0.0, l1_lambda=0.0)
sm_none.fit(X_mc_tr, y_mc_tr)
c6_test_acc_none = float(accuracy(y_mc_te, sm_none.predict(X_mc_te)))
c6_weight_norm_none = float(np.linalg.norm(sm_none.W[1:, :]))  # exclude bias row

# Part C: Ridge (L2)
sm_l2 = SoftmaxRegression(n_classes=3, learning_rate=0.5, n_epochs=1000, l2_lambda=10.0, l1_lambda=0.0)
sm_l2.fit(X_mc_tr, y_mc_tr)
c6_test_acc_l2 = float(accuracy(y_mc_te, sm_l2.predict(X_mc_te)))
c6_weight_norm_l2 = float(np.linalg.norm(sm_l2.W[1:, :]))

# Part D: Lasso (L1)
sm_l1 = SoftmaxRegression(n_classes=3, learning_rate=0.5, n_epochs=1000, l2_lambda=0.0, l1_lambda=10.0)
sm_l1.fit(X_mc_tr, y_mc_tr)
c6_test_acc_l1 = float(accuracy(y_mc_te, sm_l1.predict(X_mc_te)))
c6_n_zero_weights_l1 = int(np.sum(np.abs(sm_l1.W[1:, :]) < 1e-4))

# Part E: Comparison
c6_l2_reduced_norm = bool(c6_weight_norm_l2 < c6_weight_norm_none)
c6_l1_created_zeros = bool(c6_n_zero_weights_l1 > 0)

print(f"No reg  — Test acc: {c6_test_acc_none:.4f}, Weight L2 norm: {c6_weight_norm_none:.4f}")
print(f"L2 reg  — Test acc: {c6_test_acc_l2:.4f}, Weight L2 norm: {c6_weight_norm_l2:.4f}")
print(f"L1 reg  — Test acc: {c6_test_acc_l1:.4f}, Num zero weights: {c6_n_zero_weights_l1}")
print(f"L2 reduced norm: {c6_l2_reduced_norm}")
print(f"L1 created zeros: {c6_l1_created_zeros}")
````

````python
# ═══ Challenge 6: Grading ═══
print("Challenge 6: Multi-class Logistic Regression with Regularization")
TOTAL_SCORE += grade(c6_test_acc_none, 0.95, tolerance=0.1, points=3, name="Test acc (no reg)")
TOTAL_SCORE += grade(c6_weight_norm_none, 5.0, tolerance=10, points=2, name="Weight norm (no reg)")
TOTAL_SCORE += grade(c6_test_acc_l2, 0.90, tolerance=0.15, points=2, name="Test acc (L2)")
TOTAL_SCORE += grade(c6_weight_norm_l2, 2.0, tolerance=5, points=2, name="Weight norm (L2)")
TOTAL_SCORE += grade(c6_test_acc_l1, 0.90, tolerance=0.15, points=2, name="Test acc (L1)")
TOTAL_SCORE += grade(c6_n_zero_weights_l1, 2, tolerance=6, points=3, name="Num zero weights (L1)")
TOTAL_SCORE += grade(c6_l2_reduced_norm, True, tolerance=0, points=3, name="L2 reduced norm")
TOTAL_SCORE += grade(c6_l1_created_zeros, True, tolerance=0, points=3, name="L1 created zeros")
print()
````

### Challenge 7: Lp Regularization — Beyond L1 and L2 (15 pts)

L1 (Lasso) and L2 (Ridge) are special cases of **Lp regularization** with $p=1$ and $p=2$ respectively. The general Lp-regularized loss is:

$$J(\mathbf{w}) = \frac{1}{2n}\|\mathbf{y} - \mathbf{X}\mathbf{w}\|^2 + \lambda \sum_j |w_j|^p$$

**Task:** Implement linear regression with **Lp regularization** for arbitrary $p > 0$, and analyze how different $p$ values affect weight sparsity and model performance.

**Dataset:** Use `X_hd_tr_s` (75 samples, 20 features, standardized) and `y_hd_tr`. The true model has only 3 non-zero weights (features 0, 3, 7).

**Requirements:**

**Part A — Implementation (5 pts):**
- Implement a function `lp_regression(X, y, lam, p, learning_rate=0.01, n_epochs=5000)` that:
  - Adds a bias column to X.
  - Initializes weights to zeros.
  - Performs gradient descent with Lp regularization.
  - The gradient of $\sum_j |w_j|^p$ w.r.t. $w_j$ is: $p \cdot \text{sign}(w_j) \cdot |w_j|^{p-1}$ (use 0 when $w_j \approx 0$).
  - Do NOT regularize the bias weight $w_0$.
  - Returns the final weight vector `w`.

**Part B — Sparsity Analysis (5 pts):**
- Train with `lam=0.5` for $p \in [0.5, 1.0, 1.5, 2.0, 3.0, 4.0]$.
- For each $p$, count how many of the 20 feature weights (excluding bias) have $|w_j| < 10^{-3}$.
- Store the counts in `c7_sparsity_counts` (list of 6 integers).

**Part C — Performance Comparison (3 pts):**
- For each $p$, compute the test MSE on `X_hd_te_s` and `y_hd_te`.
- Store the test MSEs in `c7_test_mses` (list of 6 floats).
- Store the $p$ value that gives the **lowest test MSE** in `c7_best_p`.

**Part D — Understanding (2 pts):**
- Store `True` or `False` in `c7_lower_p_more_sparse` — does $p=0.5$ produce more zero weights than $p=2.0$?
- Store `True` or `False` in `c7_p1_most_sparse` — does $p=1.0$ produce the most zero weights among $p \in [1.0, 1.5, 2.0, 3.0, 4.0]$ (excluding $p=0.5$)?

**Hint:** For $p < 1$, the regularization is **non-convex** and promotes even more sparsity than L1. For $p > 2$, weights are shrunk more uniformly (less sparsity).

````python
# ═══ Challenge 7: Your code here ═══

def lp_regression(X, y, lam, p, learning_rate=0.01, n_epochs=5000):
    Xb = add_bias_column(X)
    n, d = Xb.shape
    w = np.zeros(d)
    for epoch in range(n_epochs):
        y_pred = Xb @ w
        grad = Xb.T @ (y_pred - y) / n
        # Lp regularization gradient (skip bias at index 0)
        reg_grad = np.zeros(d)
        for j in range(1, d):
            if abs(w[j]) > 1e-12:
                reg_grad[j] = p * np.sign(w[j]) * abs(w[j]) ** (p - 1)
        grad += (lam / n) * reg_grad
        w -= learning_rate * grad
    return w

# Part B: Sparsity analysis
c7_p_values = [0.5, 1.0, 1.5, 2.0, 3.0, 4.0]
c7_sparsity_counts = []
c7_test_mses = []

for p_val in c7_p_values:
    w_lp = lp_regression(X_hd_tr_s, y_hd_tr, lam=0.5, p=p_val, learning_rate=0.01, n_epochs=5000)
    n_zeros = int(np.sum(np.abs(w_lp[1:]) < 1e-3))  # exclude bias
    c7_sparsity_counts.append(n_zeros)

    # Part C: Test MSE
    Xb_te = add_bias_column(X_hd_te_s)
    test_mse = float(np.mean((y_hd_te - Xb_te @ w_lp) ** 2))
    c7_test_mses.append(test_mse)

    print(f"  p={p_val:.1f}: zeros={n_zeros}, test_mse={test_mse:.4f}, "
          f"||w||={np.linalg.norm(w_lp[1:]):.4f}")

c7_best_p = float(c7_p_values[np.argmin(c7_test_mses)])

# Part D: Understanding
c7_lower_p_more_sparse = bool(c7_sparsity_counts[0] > c7_sparsity_counts[3])  # p=0.5 vs p=2.0
# Among p in [1.0, 1.5, 2.0, 3.0, 4.0] (indices 1-5), is p=1.0 the most sparse?
c7_p1_most_sparse = bool(c7_sparsity_counts[1] == max(c7_sparsity_counts[1:]))

print(f"\nBest p (lowest test MSE): {c7_best_p}")
print(f"Sparsity counts: {c7_sparsity_counts}")
print(f"Test MSEs: {[f'{m:.4f}' for m in c7_test_mses]}")
print(f"Lower p more sparse (0.5 vs 2.0): {c7_lower_p_more_sparse}")
print(f"p=1.0 most sparse (among p>=1): {c7_p1_most_sparse}")
````

````python
# ═══ Challenge 7: Grading ═══
print("Challenge 7: Lp Regularization — Beyond L1 and L2")
TOTAL_SCORE += grade(len(c7_sparsity_counts), 6, tolerance=0, points=2, name="Sparsity counts length")
TOTAL_SCORE += grade(len(c7_test_mses), 6, tolerance=0, points=2, name="Test MSEs length")
TOTAL_SCORE += grade(c7_sparsity_counts[0], 10, tolerance=20, points=2, name="Sparsity at p=0.5 (high)")
TOTAL_SCORE += grade(c7_sparsity_counts[3], 0, tolerance=20, points=2, name="Sparsity at p=2.0 (low)")
TOTAL_SCORE += grade(c7_best_p, 0.5, tolerance=4, points=2, name="Best p value")
TOTAL_SCORE += grade(c7_lower_p_more_sparse, True, tolerance=0, points=2, name="Lower p more sparse")
TOTAL_SCORE += grade(c7_p1_most_sparse, True, tolerance=0, points=2, name="p=1 most sparse among p>=1")
print()
````
